```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import numpy as np
import matplotlib.pyplot as plt
import  warnings

warnings.filterwarnings('ignore')
```

```python
performance=pd.read_csv(r'C:\Users\suraj\Downloads\Student_Performance.csv')
performance.head()
```

# Identifying input and target field

```python
performance.columns
```

```python
performance.corr(numeric_only=True)['Performance Index']
```

```python
input_field= 'Previous Scores'
target_field= 'Performance Index'
```

## predict performance based on previous score

```python
training_data,test_data= train_test_split(performance[[input_field,target_field]],test_size=0.3,random_state=0)
```

```python
#instantiating a linear Regression model
model= LinearRegression()
```

# Training the model

```python
model.fit(training_data[[input_field]],training_data[[target_field]])
```

# Making Predictions

```python
model.predict(test_data[[input_field]])
```

# Evaluating the model

```python
from sklearn.metrics import mean_absolute_error
print(f'mean absolute erorr: {mean_absolute_error(test_data[target_field],model.predict(test_data[[input_field]]))}')
```

```python
from sklearn.metrics import mean_squared_error
print(f'mean squared erorr: {mean_squared_error(test_data[target_field],model.predict(test_data[[input_field]]))}')
```

```python
from sklearn.metrics import root_mean_squared_error
print(f'root mean squared  erorr: {root_mean_squared_error(test_data[target_field],model.predict(test_data[[input_field]]))}')
```

```python
from sklearn.metrics import r2_score
print(f' r2 score: {r2_score(test_data[target_field],model.predict(test_data[[input_field]]))}')
```

```python
model_2= LinearRegression(positive=True)
model_2.fit(training_data[[input_field]],training_data[[target_field]])
print(f'mean absolute erorr: {mean_absolute_error(test_data[target_field],model_2.predict(test_data[[input_field]]))}')
print(f'mean squared erorr : {mean_squared_error(test_data[target_field],model_2.predict(test_data[[input_field]]))}')
print(f'root mean squared  erorr: {root_mean_squared_error(test_data[target_field],model_2.predict(test_data[[input_field]]))}')
print(f'r2 score: {r2_score(test_data[target_field],model_2.predict(test_data[[input_field]]))}')
```

```python
def adjusted_r2(actual,prediction,inputs):
    return ((1-r2_score(actual,prediction))* (len(actual)-1))/ (len(actual)-inputs-1)
adjusted_r2(test_data[target_field],model_2.predict(test_data[[input_field]]),model_2.n_features_in_)
```

# Linear Regression using  multiple  Input fields(Multiple Linear Regression)

```python
input_fields = ['Hours Studied', 'Previous Scores',
                'Sleep Hours', 'Sample Question Papers Practiced']

training_data, test_data = train_test_split(
    performance[input_fields + [target_field]],
    test_size=0.3,
    random_state=0
)

model_3 = LinearRegression()

model_3.fit(
    training_data[input_fields],
    training_data[target_field]
)

print(f'mean absolute error: {mean_absolute_error(test_data[target_field], model_3.predict(test_data[input_fields]))}')
print(f'mean squared error: {mean_squared_error(test_data[target_field], model_3.predict(test_data[input_fields]))}')
print(f'root mean squared error: {root_mean_squared_error(test_data[target_field], model_3.predict(test_data[input_fields]))}')
print(f'r2 score: {r2_score(test_data[target_field], model_3.predict(test_data[input_fields]))}')
```

# Saving model for future use

```python
import joblib as jb

#save model
linear_regression_model={'inputs':input_fields,'target': target_field,'model': model_3}
jb.dump(linear_regression_model,'linear_regression.joblib')
```

```python
saved_model= jb.load('linear_regression.joblib')

input_values={}
for feature in saved_model['inputs']:
    value= input(f'enter the {feature.lower()}:')
    input_values.update({feature:float(value)})

print(f'predicted performance:{saved_model['model'].predict(pd.DataFrame(input_values,index=[0]))}')
```
